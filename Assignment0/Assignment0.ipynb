### Part 3: Some manipulations (6 points)

```python
print('Question 4')
print('What is the difference between the value of dL/dw and what it “should" be, according to the notebook output?')

tkwargs = {'dtype': torch.float, 'device': device } # For tensor initialization
w = torch.tensor (1., requires_grad=True, ** tkwargs )
b = torch.tensor (0., requires_grad=True, ** tkwargs )
x = torch.tensor ([0.5], ** tkwargs )
y = torch.tensor ([6.8], ** tkwargs )
z = w * x + b
loss = (y - z). pow (2).sum ()
loss.backward ()
print (f'The squared error is { loss.item ()}')
print (' ---------------------')
print (f'dL/dw = {w. grad }, should be {2* x*(z - y)}')
print (f'dL/db = {b. grad }, should be {2*( z - y)}')
print()
```

```python
print('Question 5')
print('What is the “new" squared error?') 
print('How much lower is it than the original squared error in the previous cell?')

alpha = 0.01 # Learning rate
with torch.no_grad ():
    w = w - alpha * w. grad
    b = b - alpha * b. grad
print(f'The new value of w is {w. item ()}')
print(f'The new value of b is {b. item ()}')
print('---------------------')
z = w * x + b
loss = (y - z). pow (2).sum ()
print(f'The new squared error is { loss.item ()}')
print()
```
